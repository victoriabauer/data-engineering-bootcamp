How to open a file:
The file path (file directory+ file name) is the first argument of open
2nd parameter is the mode, in this case r = reading
w = writing
a = appending
file1 = file object

```python
file1 = open("/resources/data/Example1.txt","r")
file1.name #returns the file path
file1.mode #returns the attribute
file1.close()
```

Use with statement as it automatically closes file after opened

```python
with open("Example1.txt","r") as file1:
    file_stuff = file1.read()
    print(file_stuff)
print(file1.closed) #you can't read anymore of the file
print(file_stuff) #but you can print the file content as file_stuff
```

```python
with open("Example1.txt","r") as file1:
    file_stuff = file1.readline()
    print(file_stuff)
```

Each time you call readline(), a new line will be read. You can use a loop to read reach line.

```python
with open("Example1.txt","r") as file1:
    file_stuff = file1.readline(16)
    print(file_stuff)
    file_stuff = file1.readline(5)
    print(file_stuff)
    file_stuff = file1.readline(9)
    print(file_stuff)

#This will read up to n characters, and will keep grabbing characters even if they're on the same line
```

If you want to read characters from a specific position in the file, you can use the seek() method. This method moves the file pointer (like a cursor) to a particular position. The position is specified in bytes, so you'll need to know the byte offset of the characters you want to read.

```python
file.seek(10)  # Move to the 11th byte (0-based index)
```
