Enumerate: Use this if you want to keep track of index and corrosponding element in a collection

```python
def print_stuff(Stuff):
    for i,s in enumerate(Stuff):
        print("Album",i,"Rating is",s)

album_ratings = [10,8,5]
print_stuff(album_ratings)
```
