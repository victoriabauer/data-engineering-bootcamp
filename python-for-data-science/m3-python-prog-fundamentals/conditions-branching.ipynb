If statements: They allow code to branch--choose one course of action or another

```python
age = 8

if(age>=18):
    print("You may enter")

print("Move on") #This line will happen regardless
```

```python
#Else
age = 8

if(age>=18):
    print("You may enter")
else:
    print("Go see barney")

print("Move on") #This line will happen regardless
```

```python
#Elif
age = 8

if(age>=18):
    print("You may enter")
elif(age>=15):
    print("Go see linkin park")

print("Move on, you're too young") #This line will happen regardless
```

Logic operators
They take boolean operands and produce boolean values
x and x => only produces a true when both operands are true
x or x => only produces a false when both operands are false
not(x) =>produces the opposite boolean

```python
x = 0
if x < 10 or x > 19:
    print("x is single digit or 20s plus")
else:
    print("x is teens")
```
