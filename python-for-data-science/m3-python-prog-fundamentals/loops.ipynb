```python
#loops
N = 5
range(N) #outputs an ordered sequence as a list i [0,1,2,3,4]
#if input is a positive integer then the output is a sequence with length N.
range(10,15) #=>[10,11,12,13,14], output does not include last number

#for loops
x=0
for i in range(5):
    x+=1
    print(x,i)
```

```python
squares = ["red","yellow","green"]
for square in squares:
    print(square)
```

```python
#Here is the syntax to iterate through a list and provide the index of each element
for i,square in enumerate(squares):
    print(i,square)
```

```python
#while loops
#runs a loop while a condition is met
#as opposed to for loop which runs a set number of times
squares = ['orange','orange','purple','orange','blue']
new_squares = []
i=0
while(squares[i]=='orange'):
    new_squares.append(squares[i])
    i+=1 #or i=i+1
print(new_squares)
```

Use for loop when number of iterations is known and need to evaluate a sequence of elements

Use while loops when number of iterations is not known or when you're waiting for a specific condition to be met

Break statement:
This immediately terminates the loop it appears in, regardless of loop conditions. It's commonly used when a specific condition is met and no further iterations are needed. 

```python
for num in range(1,11):
    if num == 5:
        print("Breaking the loop at:",num)
        break
    print(num)
```

Continue statement:
skips the current iteration and moves to the next one in the loop. It's used when you want to ignore certain conditions but continue looping. 

```python
for num in range(1,7):
    if num == 3:
        continue
    print(num)
```
