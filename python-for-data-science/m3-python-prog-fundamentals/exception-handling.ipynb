Exception Handling:
Used for asking user input. Need to know how to handle certain cases.

```python
try:
    getfile = open('myfile','r')
    getfile.write('File for exception handling')
except IOError:
    print("Unable to open or read the data in the file.")
except:
    print("Some other error occurred!")
else:
    print("File was written successfully!")
finally:
    getfile.close()
    print('File is now closed.')
```

Note: You want to try to define the errors so that when programmer goes back to debug, they'll know the specific issue. Below are some examples of exceptions:

```python
# ZeroDivisionError demo
try:
    1/0
except Exception as e:
    print(type(e).__name__, "-", e)
```

```python
# NameError demo
try:
    y = a + 5  # a is undefined
except Exception as e:
    print(type(e).__name__, "-", e)
```

```python
# IndexError demo
try:
    a = [1, 2, 3]
    a[10]
except Exception as e:
    print(type(e).__name__, "-", e)
```

```python
a = 1

try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
    print("Success a=",a)
except ZeroDivisionError:
    print("The number you provided cant divide 1 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
        
```
