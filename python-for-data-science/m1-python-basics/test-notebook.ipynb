The error message tells you:

- where the error occurred (more useful in large notebook cells or scripts), and
- what kind of error it was (NameError)

Python is an interpreter language. Compiled languages will analyze whole program to catch any errors, while python will execute line by line. When it runs into an error it will stop the program unless the error is handled by the user in code.
