Regular Expressions or RegEx: A tool for matching and handling strings
Functions: search, split, findall, and sub

```python
import re
```

```python
s1 = 'The BodyGuard is the best album'

# Defines the pattern
pattern = r"Body"

result = re.search(pattern,s1)

if result:
    print('Match!')
else:
    print('No match!')
```
