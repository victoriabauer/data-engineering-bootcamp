Regular Expressions or RegEx: A tool for matching and handling strings
Functions: search, split, findall, and sub

```python
import re
```

```python
s1 = 'The BodyGuard is the best album'

# Defines the pattern
pattern = r"Body"

result = re.search(pattern,s1)

if result:
    print('Match!')
else:
    print('No match!')
```

Special Sequence        Meaning                                                             Example
\d              Matches any digit character (0-9)                                           "123" matches "\d\d\d"
\D              Matches any non-digit character                                             "hello" matches "\D\D\D\D\D"
\w              Matches any word character (a-z, A-Z, 0-9, and _)                           "hi_world" matches "\w\w\w\w\w\w\w\w"
\W              Matches any non-word character                                              "@#$%" matches "\W\W\W\W"
\s              Matches any whitespace character (space, tab, newline, etc.)                "hi world" matches "\w\w\s\w\w\w\w\w"
\S              Matches any non-whitespace character                                        "hi_world" matches "\S\S\S\S\S\S\S\S"
\b              Matches the boundary between a word character and a non-word character      "cat" matches "\bcat\b" in "The cat sat on the mat"
\B              Matches any position that is not a word boundary                            "cat" matches "\Bcat\B" in "category" but not in "The cat sat on the mat"

findall(): finds all occurances of a pattern in a string, returns a list of the matches found

```python
pattern = r"\W" #matches non-word characters
text = "Hello, world!"
matches = re.findall(pattern,text)
print("Matches:",matches)
```

You can also use findall for substrings instead of patterns

```python
s2 = "The BodyGuard is the best album of 'Whitney Houston'."
result = re.findall('st',s2)
print(result)
# If you want the number of occurences
print(len(result))
```

split(): a function that splits a string into an array of substrings based on a specified pattern

```python
split_array = re.split(r"\s",s2)
print(split_array)
```

sub function: used to replace all occurences of a pattern in a string with a specified replacement

```python
pattern = r"Whitney Houston"
replacement = 'legend'
new_str = re.sub(pattern,replacement,s2,flags=re.IGNORECASE) #re.IGNORECASE makes the search case-insensitive, so it matches whitney houston in any letter case
print(new_str)
```
