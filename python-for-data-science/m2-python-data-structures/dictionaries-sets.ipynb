Dictionaries
A type of collection in python
Similar to a list, however dictionaries have key's and values.
Key is like the index but it does not have to be an integer.
{} = dictionary creation
The values can be immutable, mutable, and duplicates
dict = {"key1":1,"key2":"2","key3":[3,3,3],"key4":(4,4,4),('key5'):5}

Observation: ('key5'):5

dict["key1"]=>"1"
