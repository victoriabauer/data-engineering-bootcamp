Set: an unordered collection of unique objects
Notice below that when you print the set, only unique elements are printed


```python
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
print(set1)
#print("hello")
```

You can also make a set from a list

```python
list = [1,1,1,2,3]
set4 = set(list)
print(set4)
```

Logic Operations

```python
# Sample Sets

album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])

#Below return sets
intersection = album_set1 & album_set2

#Keep elements in set1 not in both or set2
album_set1.difference(album_set2)  

#Keep elements in set2 not in both or set1
album_set2.difference(album_set1) 

#same thing as intersection
album_set1.intersection(album_set2)  

#Finds all elements between set1 and set2
album_set1.union(album_set2) 

#Below return True or False

#Check if one set is a superset of another. The issuperset() method is used to verify whether the current set 
# contains all elements of another set. It returns True if every element of the specified set exists within the current set; 
# otherwise, it returns False.
set(album_set1).issuperset(album_set2) 

set(album_set2).issubset(album_set1) 
set({"Back in Black", "AC/DC"}).issubset(album_set1) 
album_set1.issuperset({"Back in Black", "AC/DC"}) 
```
