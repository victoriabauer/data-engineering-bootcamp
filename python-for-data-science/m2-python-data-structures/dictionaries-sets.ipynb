```python
# Sample Sets

album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])

#Below return sets
intersection = album_set1 & album_set2

#Keep elements in set1 not in both or set2
album_set1.difference(album_set2)  

#Keep elements in set2 not in both or set1
album_set2.difference(album_set1) 

#same thing as intersection
album_set1.intersection(album_set2)  

#Finds all elements between set1 and set2
album_set1.union(album_set2) 

#Below return True or False

#Check if one set is a superset of another. The issuperset() method is used to verify whether the current set 
# contains all elements of another set. It returns True if every element of the specified set exists within the current set; 
# otherwise, it returns False.
set(album_set1).issuperset(album_set2) 

set(album_set2).issubset(album_set1) 
set({"Back in Black", "AC/DC"}).issubset(album_set1) 
album_set1.issuperset({"Back in Black", "AC/DC"}) 
```

The issue with my solution is that the while loop might add the line before if it checks there's something in the line. The other issue is that. A dictionary stores each key only once, so if a word appears multiple times in B, you won't print its results multiple times.
For example, if B = ['a', 'c', 'a'], your output needs three lines, even though 'a' appears twice in B.
Instead, loop through B itself and print the corresponding dictionary value for each word. Below is another approach.

Also, rather than having five million loops, the code only uses two. One to store A. One to check A and print A if it matches anything form B.

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT


from collections import defaultdict
import sys

first_line = sys.stdin.readline().strip().split()
a_len = int(first_line[0])
b_len = int(first_line[1])
d = defaultdict(list)

for i in range(a_len):
    word = sys.stdin.readline().strip()
    d[word].append(i+1) #+1 for position

for i in range(b_len):
    word = sys.stdin.readline().strip()
    if d[word]:
        print(*d[word])
    else:
        print(-1)
    
            

    
    
```
