Tuples and lists
- compound data structure

Tuples: an ordered sequence
- content inside paranthesis and comma separated
- Ex: Ratings = (1,2,3,4)
- Content can be int, float, str, etc. and can mixed
- variable type = tuple
- Content of tuple can accessed via an index, for ex: Ratings[0] => 1
- like strings, you can access tuple with negative index with -1 = last letter, first letter will be the larger negative number

You can combine tuples by adding them

```python
t1 = (1,2)
t3 = (3,4)
print(t1+t3)
```

Slicing tuples
Note: the 2nd number is always 1 + target index

```python
t2 = (4,5,6,7,8)
t2[0:3]
```

Tuples are immutable: You can't manipulate the individual contents/the original llst

Lists:
Like tuples are immutable
Lists can hold basic data types, other lists, and tuples
Access is same with pos or neg indexes
Concat is the same
l1[0,3] => will get you first part of the list
You can extend the list

```python
L = [1,2]
L.extend([3,4])
print(L)
```

```python
L = [1,2]
L.append([3,4])
print(L)
```

Aliasing
When two names/variables refer to the same list. If you change an element, both lists are impacted because they refer to the same one.

```python
A = [1,2,3,4]
B = A
A[0] = 'apple'
print(A)
print(B)
```

To get around this affect, you can clone lists

```python
A = [1,2,3,4]
B = A[:]
A[0] = 'apple'
print(A)
print(B)
```
